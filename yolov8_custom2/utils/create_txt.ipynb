```python
import pandas as pd
import os

info_label = {
    'age': ['20-30s', '40-50s', 'Baby', 'Kid', 'Senior', 'Teenager'],
    'race': ['Caucasian', 'Mongoloid', 'Negroid'],
    'masked': ['masked', 'unmasked'],
    'skintone': ['dark', 'light', 'mid-dark', 'mid-light'],
    'emotion': ['Anger', 'Disgust', 'Fear', 'Happiness', 'Neutral', 'Sadness', 'Surprise'],
    'gender': ['Female', 'Male']
}

all_class = []
for key in info_label.keys():
    all_class += info_label[key]
all_class.insert(0, 'face')

def convert_bbox_to_yolo(bbox, img_width, img_height):
    x_min, y_min, width, height = bbox
    x_center = x_min + width / 2
    y_center = y_min + height / 2

    # Normalize by image size
    return [x_center / img_width, y_center / img_height, width / img_width, height / img_height]

def save_yolo_format(file_name, bbox, class_id, save_dir):
    yolo_data = f"{class_id} {' '.join(map(str, bbox))}\n"
    file_path = os.path.join(save_dir, file_name.replace('.jpg', '.txt'))

    with open(file_path, 'a') as file:
        file.write(yolo_data)

# Load CSV file
csv_file = '../csvfile/private_labels.csv'
data = pd.read_csv(csv_file)

# Directory to save .txt files
save_directory = r"D:\Pixta\private_test\labels"
os.makedirs(save_directory, exist_ok=True)

# Process each row
ls = []
for index, row in data.iterrows():
    bbox = eval(row['bbox']) # Assuming bbox is stored as a string representation of a list
    yolo_bbox = convert_bbox_to_yolo(bbox, row['width'], row['height'])
    # if row['file_name'] in ls:
    #     print(row['file_name'])
    # ls.append(row['file_name'])
    class_id = [row[key] for i,key in enumerate(row.keys()) if i >= 4]
    class_id.insert(0, 'face')

    # print(class_id)
    for cid in class_id:
        ind_cid = all_class.index(cid)
        save_yolo_format(row['file_name'], yolo_bbox, ind_cid, save_directory)

print("Conversion to YOLO format completed.")
```
